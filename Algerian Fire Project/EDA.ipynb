## Data Cleaning

```python
## missing values

dataset.isnull().sum()
```

```python
dataset[dataset.isnull().any(axis=1)]
```

#### The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region

```python
# WKT that the Sidi-Bel Abbes starts from 122. So we create 2 columns for both the datasets in the form on 0 and 1

dataset.loc[:122, 'Region'] = 0
dataset.loc[122:, 'Region'] = 1
df = dataset
```

```python
df.info()
```

```python
df[['Region']] = df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# remove the 122 row with null values

df = df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum() # Now, zero null values
```

```python
df.iloc[[122]]
```

```python
## drop this 122nd row

df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]] # done
```

```python
df.columns
```

```python
# fix spaces in column names
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
## change the required columns to int data type

# change : 'day', 'month', 'year', 'Temperature', 'RH', 'Ws' into int

df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']] = df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int) 
```

```python
df.info() # checking if converetd to int
```

```python
## changing the other columns to float data type

objects = [features for features in df.columns if df[features].dtypes == 'O']

# other columns that has 'O' that is Object as data type.

# can also be written as 
'''
object = []
for features in df.columns:
    if df[features] == 'O':
        object.append(features)
'''
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)

# except for classes column, we changed others to float
```

```python
## Lets save the Cleaned Dataset

df.to_csv('Algerian_forest_fires_dataset_cleaned.csv', index=False)
```

```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'), 'not fire', 'fire')
```

```python
## Monthly Fire Analysis
dftemp = df.loc[df['Region']==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue='Classes', data=dftemp)
plt.ylabel('Number of Fires', weight='bold')
plt.xlabel('Months', weight='bold')
plt.title('Fire Analysis of Sidi-Bel Regions', weight='bold')
plt.savefig('Fire_Analysis_of_Sidi-Bel_Region.png')
plt.show()
```

```python
## Monthly Fire Analysis for Bejaia Region
dftemp = df.loc[df['Region']==0]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue='Classes', data=dftemp)
plt.ylabel('Number of Fires', weight='bold')
plt.xlabel('Months', weight='bold')
plt.title('Fire Analysis of Bejaia Regions', weight='bold')
plt.savefig('Fire_Analysis_of_Bejaia_Region.png')
plt.show()
```

Its observed that August and September had the most number of forest fires for both regions. And from the above plot of months, we can understand few things

Most of the fires happened in August and very high Fires happened in only 3 months - June, July and August.

Less Fires was on September
